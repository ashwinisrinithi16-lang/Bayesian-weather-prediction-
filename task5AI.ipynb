Bayes theorem

```python
total_days = 20

rain = 8
not_rain = 12

cloudy_rain = 7
cloudy_not_rain = 3

p_rain = rain / total_days
p_not_rain = not_rain / total_days

p_cloudy_rain = cloudy_rain / rain
p_cloudy_not_rain = cloudy_not_rain / not_rain

p_cloudy = (p_cloudy_rain * p_rain) + (p_cloudy_not_rain * p_not_rain)

p_rain_cloudy = (p_cloudy_rain * p_rain) / p_cloudy
p_not_rain_cloudy = (p_cloudy_not_rain * p_not_rain) / p_cloudy

print("BAYESIAN WEATHER PREDICTION")
print("---------------------------")

print("P(Rain) =", round(p_rain, 2))
print("P(Not Rain) =", round(p_not_rain, 2))
print("P(Cloudy | Rain) =", round(p_cloudy_rain, 2))
print("P(Cloudy | Not Rain) =", round(p_cloudy_not_rain, 2))
print("P(Rain | Cloudy) =", round(p_rain_cloudy, 2))
print("P(Not Rain | Cloudy) =", round(p_not_rain_cloudy, 2))

if p_rain_cloudy > p_not_rain_cloudy:
    print("Predicted Weather = Rain")
else:
    print("Predicted Weather = Not Rain")
```

Bayes' Theorem is used to find the probability of rain when the sky is cloudy:

Formula:

P(Rain | Cloudy) =
P(Cloudy | Rain) × P(Rain)
--------------------------------
P(Cloudy)

Similarly:

P(Not Rain | Cloudy) =
P(Cloudy | Not Rain) × P(Not Rain)
-----------------------------------
P(Cloudy)

Here:

P(Rain | Cloudy) = 0.82
P(Not Rain | Cloudy) = 0.18

Since 0.82 > 0.18, the predicted weather condition is Rain.
